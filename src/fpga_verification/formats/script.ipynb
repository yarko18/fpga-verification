```python
from fpga_verification.formats import QFormat, UIntFormat
```

```python
assert QFormat(3, 0, signed=False).qraw_to_int(0b111) == 7
assert QFormat(3, 0, signed=True).qraw_to_int(0b111) == -1
```

```python
custom_q = QFormat(2, 14, signed=False)

raw = custom_q.float_to_qraw(2.0)
print( raw )
assert raw == 2 << 14
```

```python
raw = custom_q.float_to_qraw(4.0)
print( custom_q.qraw_to_float( raw ) )
assert custom_q.qraw_to_float( raw ) != 4
```

```python
gain_q = QFormat(2, 14, signed=False)
assert gain_q.ones() == 16384  # raw representation of 1.0
```
